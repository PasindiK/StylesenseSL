```python
import os
import json
import yaml
from datetime import datetime

# Paths
data_mesh_path = "/Users/nandunmadawa/Desktop/STYLESENSE/data-mesh/Data_Mesh_Domains"
metadata_file = os.path.join(data_mesh_path, "metadata/data_mesh_metadata.json")
contracts_path = os.path.join("/Users/nandunmadawa/Desktop/STYLESENSE/data-mesh/Contracts")

# Make contracts folder if it doesn't exist
os.makedirs(contracts_path, exist_ok=True)

# Load metadata
with open(metadata_file, "r") as f:
    metadata = json.load(f)

# Generate contracts for each domain
for domain_info in metadata:
    domain_name = domain_info['domain_name']
    domain_contract = {
        'domain_name': domain_name,
        'owner': 'nadunthenuwara@gmail.com',  # Replace with actual owner
        'description': f'Auto-generated contract for {domain_name}. Add full description here.',
        'schema': [],
        'update_frequency': 'weekly',  # You can change per domain
        'sla': {
            'freshness': 'Data should be updated weekly',
            'completeness': 'No missing primary key values allowed'
        },
        'examples': [
            f"SELECT * FROM {domain_name} LIMIT 10"
        ]
    }

    # Add column details
    for col in domain_info['data_types']:
        domain_contract['schema'].append({
            'column': col,
            'type': str(domain_info['data_types'][col]),
            'nullable': bool(domain_info['nullable_columns'].get(col, True)),
            'description': 'Add column description here'
        })

    # Write YAML contract
    contract_file = os.path.join(contracts_path, f"{domain_name}.yml")
    with open(contract_file, "w") as f:
        yaml.dump(domain_contract, f, sort_keys=False)

    print(f"✅ Contract generated: {contract_file}")
```
