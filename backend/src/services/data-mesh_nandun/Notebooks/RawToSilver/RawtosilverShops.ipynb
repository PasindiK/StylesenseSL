# Raw to Silver: Data Preprocessing
This notebook demonstrates how to preprocess transaction data from the raw layer and save a cleaned version to the silver layer. You can adapt this workflow for other raw datasets in your project.

```python


import pandas as pd

RAW_PATH = "../raw-data/shops_dataset.csv"
SILVER_PATH = "../Silver-data/shops_clean.csv"

# Load raw shops data
shopss = pd.read_csv(RAW_PATH)
print("Raw shops Shape:", shopss.shape)
shopss.head()
```

## Step 1: Remove Duplicates

```python
# Remove duplicate rows
shopss = shopss.drop_duplicates()
print("After removing duplicates:", shopss.shape)
```

## Step 2: Handle Missing Values

```python
# Check for missing values
missing_summary = shopss.isnull().sum()
print("Missing values per column:\n", missing_summary)
```

```python


# shop_name
shopss['shop_name'] = shopss['shop_name'].fillna('UNKNOWN_SHOP')

# district (fill from location / city)
shopss['district'] = shopss['district'].fillna(shopss['location'])

# is_active (boolean)
shopss['is_active'] = shopss['is_active'].fillna(True)

# operating hours
shopss['operating_hours_open'] = shopss['operating_hours_open'].fillna('09:00')
shopss['operating_hours_close'] = shopss['operating_hours_close'].fillna('21:00')




print("Missing values per column AFTER cleaning:\n", shopss.isnull().sum())


print("After Cleaned missings:", shopss.shape)



```

```python
shopss.head(20)
```

## Step 3: Data Type Corrections (if needed)

```python


shopss['phone_number'] = shopss['phone_number'].astype(str)


```

```python
shopss.dtypes
```

## Step 4: Save Cleaned Data to Silver Layer

```python
# Save the cleaned dataframe to the silver layer
import os
os.makedirs(os.path.dirname(SILVER_PATH), exist_ok=True)
shopss.to_csv(SILVER_PATH, index=False)
print(f"Cleaned shop data saved to {SILVER_PATH}")
```
