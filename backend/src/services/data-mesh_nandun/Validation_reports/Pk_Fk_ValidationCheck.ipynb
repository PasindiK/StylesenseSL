```python
import pandas as pd

base_path = "../Data_Mesh_Domains"

users = pd.read_csv(f"{base_path}/users_domain/users_domain.csv")
products = pd.read_csv(f"{base_path}/product_domain/product_domain.csv")
shops = pd.read_csv(f"{base_path}/shop_domain/shop_domain.csv")
sales = pd.read_csv(f"{base_path}/sales_domain/sales_domain.csv")
interactions = pd.read_csv(f"{base_path}/interaction_domain/interaction_domain.csv")
```

```python
def check_primary_key(df, pk_name, domain):
    print(f"\n🔍 {domain} - PK Check ({pk_name})")
    print("Nulls:", df[pk_name].isna().sum())
    print("Duplicates:", df[pk_name].duplicated().sum())

check_primary_key(users, "user_id", "Users Domain")
check_primary_key(products, "product_id", "Product Domain")
check_primary_key(shops, "shop_id", "Shop Domain")
check_primary_key(sales, "transaction_id", "Sales Domain")
check_primary_key(interactions, "interaction_id", "Interaction Domain")
```

```python
def check_foreign_key(child_df, child_col, parent_df, parent_col, description):
    missing = ~child_df[child_col].isin(parent_df[parent_col])
    print(f"\n {description}")
    print("Missing references:", missing.sum())

# Sales → Users
check_foreign_key(sales, "user_id", users, "user_id", "Sales.user_id → Users.user_id")

# Sales → Products
check_foreign_key(sales, "product_id", products, "product_id", "Sales.product_id → Products.product_id")

# Sales → Shops
check_foreign_key(sales, "shop_id", shops, "shop_id", "Sales.shop_id → Shops.shop_id")

# Interactions → Users
check_foreign_key(interactions, "user_id", users, "user_id", "Interactions.user_id → Users.user_id")

# Interactions → Products
check_foreign_key(interactions, "product_id", products, "product_id", "Interactions.product_id → Products.product_id")
```
