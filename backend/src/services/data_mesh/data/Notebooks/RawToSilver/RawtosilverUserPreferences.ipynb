# Raw to Silver: Data Preprocessing
This notebook demonstrates how to preprocess transaction data from the raw layer and save a cleaned version to the silver layer. You can adapt this workflow for other raw datasets in your project.

```python


import pandas as pd

RAW_PATH = "../raw-data/user_preferences_dataset.csv"
SILVER_PATH = "../Silver-data/users_preferences_clean.csv"

# Load raw users data
usersPreferences = pd.read_csv(RAW_PATH)
print("Raw Users Shape:", usersPreferences.shape)
usersPreferences.head()
```

## Step 1: Remove Duplicates

```python
# Remove duplicate rows
usersPreferences = usersPreferences.drop_duplicates()
print("After removing duplicates:", usersPreferences.shape)
```

## Step 2: Handle Missing Values

```python
# Check for missing values
missing_summary = usersPreferences.isnull().sum()
print("Missing values per column:\n", missing_summary)
```

```python
# Drop rows where t_id is missing
usersPreferences = usersPreferences.dropna(subset=['preference_id'])
usersPreferences = usersPreferences.dropna(subset=['user_id'])
# Fill missing columns
usersPreferences['preferred_categories'] = usersPreferences['preferred_categories'].fillna(usersPreferences['preferred_categories'].mode()[0])
usersPreferences['preferred_colors'] = usersPreferences['preferred_colors'].fillna(usersPreferences['preferred_colors'].mode()[0])
usersPreferences['preferred_fabrics'] = usersPreferences['preferred_fabrics'].fillna(usersPreferences['preferred_fabrics'].mode()[0])
usersPreferences['preferred_brands'] = usersPreferences['preferred_brands'].fillna(usersPreferences['preferred_brands'].mode()[0])
usersPreferences['preferred_styles'] = usersPreferences['preferred_styles'].fillna(usersPreferences['preferred_styles'].mode()[0])
usersPreferences['skin_tone'] = usersPreferences['skin_tone'].fillna(usersPreferences['skin_tone'].mode()[0])
usersPreferences['body_type'] = usersPreferences['body_type'].fillna(usersPreferences['body_type'].mode()[0])
usersPreferences['price_sensitivity'] = usersPreferences['price_sensitivity'].fillna(usersPreferences['price_sensitivity'].mode()[0])

usersPreferences['updated_ts'] = usersPreferences['updated_ts'].fillna(pd.Timestamp.now())
```

```python
usersPreferences.head()
```

## Step 3: Data Type Corrections (if needed)

```python
usersPreferences.dtypes
```

```python
usersPreferences['preference_id'] = usersPreferences['preference_id'].astype(int)
usersPreferences['user_id'] = usersPreferences['user_id'].astype(int)
```

## Step 4: Save Cleaned Data to Silver Layer

```python
# Save the cleaned dataframe to the silver layer
import os
os.makedirs(os.path.dirname(SILVER_PATH), exist_ok=True)
usersPreferences.to_csv(SILVER_PATH, index=False)
print(f"Cleaned user preferences data saved to {SILVER_PATH}")
```
