# Raw to Silver: Data Preprocessing
This notebook demonstrates how to preprocess transaction data from the raw layer and save a cleaned version to the silver layer. You can adapt this workflow for other raw datasets in your project.

```python


import pandas as pd

RAW_PATH = "../raw-data/trends_dataset.csv"
SILVER_PATH = "../Silver-data/trends_clean.csv"

# Load raw products data
trends = pd.read_csv(RAW_PATH)
print("Raw trends Shape:", trends.shape)
trends.head()
```

## Step 1: Remove Duplicates

```python
# Remove duplicate rows
trends = trends.drop_duplicates()
print("After removing duplicates:", trends.shape)
```

## Step 2: Handle Missing Values

```python
# Check for missing values
missing_summary = trends.isnull().sum()
print("Missing values per column:\n", missing_summary)
```

```python
# Drop rows where product_id is missing
trends = trends.dropna(subset=['trend_id'])

# Fill missing columns
trends['trend_name'] = trends['trend_name'].fillna('Unknown')
trends['trend_tags'] = trends['trend_tags'].fillna('Unknown')
trends['emerging_designers'] = trends['emerging_designers'].fillna('Unknown')
trends['trend_category'] = trends['trend_category'].fillna(trends['trend_category'].mode()[0])
trends['trend_score'] = trends['trend_score'].fillna(trends['trend_score'].median())
trends['week'] = trends['week'].fillna(trends['week'].mode()[0])
trends['year'] = trends['year'].fillna(trends['year'].mode()[0])


```

## Step 3: Data Type Corrections (if needed)

```python
trends.dtypes
```

```python
trends['trend_id'] = trends['trend_id'].astype(int)

```

## Step 4: Save Cleaned Data to Silver Layer

```python
# Save the cleaned dataframe to the silver layer
import os
os.makedirs(os.path.dirname(SILVER_PATH), exist_ok=True)
trends.to_csv(SILVER_PATH, index=False)
print(f"Cleaned products data saved to {SILVER_PATH}")
```
