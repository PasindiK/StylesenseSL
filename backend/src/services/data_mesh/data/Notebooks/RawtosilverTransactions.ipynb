# Raw to Silver: Data Preprocessing
This notebook demonstrates how to preprocess transaction data from the raw layer and save a cleaned version to the silver layer. You can adapt this workflow for other raw datasets in your project.

```python


import pandas as pd

RAW_PATH = "../raw-data/transactions10K.csv"
SILVER_PATH = "../Silver-data/transactions_clean.csv"

# Load raw transactions data
transactions = pd.read_csv(RAW_PATH)
print("Raw Transactions Shape:", transactions.shape)
transactions.head()
```

## Step 1: Remove Duplicates

```python
# Remove duplicate rows
transactions = transactions.drop_duplicates()
print("After removing duplicates:", transactions.shape)
```

## Step 2: Handle Missing Values

```python
# Check for missing values
missing_summary = transactions.isnull().sum()
print("Missing values per column:\n", missing_summary)
```

```python
# ---------------------------
# Drop rows with critical nulls
# ---------------------------
critical_cols = [
    'shop_id',
    'unit_price',
    'total_amount',
    'discount_amount',
    'tax_amount',
    'final_amount',
    'transaction_date',
    'transaction_ts',
    'delivery_address',
    'delivery_date'
]

transactions = transactions.dropna(subset=critical_cols)

missing_summary = transactions.isnull().sum()
print("Missing values per column:\n", missing_summary)

print("After Cleaned missings:", transactions.shape)




```

```python
transactions.head()
```

## Step 3: Data Type Corrections (if needed)

```python
# Example: Convert date columns to datetime
if 'transaction_date' in transactions.columns:
    transactions['transaction_date'] = pd.to_datetime(transactions['transaction_date'], errors='coerce')
    
```

## Step 4: Save Cleaned Data to Silver Layer

```python
# Save the cleaned dataframe to the silver layer
import os
os.makedirs(os.path.dirname(SILVER_PATH), exist_ok=True)
transactions.to_csv(SILVER_PATH, index=False)
print(f"Cleaned transactions data saved to {SILVER_PATH}")
```
