```python
import os
import shutil
import pandas as pd

# You are running from:
# data-mesh/Notebooks/SilverToDataMesh

SILVER_PATH = "../../Data/Silver-data"
DATA_MESH_PATH = "../../Data_Mesh_Domains"

print("Silver Path:", os.path.abspath(SILVER_PATH))
print("Data Mesh Path:", os.path.abspath(DATA_MESH_PATH))
```

```python
# -----------------------------
# DOMAIN FOLDER MAP
# -----------------------------
# Map each dataset to its target domain folder
dataset_domain_map = {
    "interactions_clean.csv": "customer_domain",
    "products_clean.csv": "product_domain",
    "shops_clean.csv": "shop_domain",
    "transactions_clean.csv": "sales_domain",
    "trends_clean.csv": "engagement_domain",
    "users_clean.csv": "users_domain",
    "users_preferences_clean.csv": "user_preferences_domain"
}

# -----------------------------
# MOVE DATASETS
# -----------------------------
for dataset_name, domain_folder in dataset_domain_map.items():
    source_file = os.path.join(SILVER_PATH, dataset_name)
    target_folder = os.path.join(DATA_MESH_PATH, domain_folder)
    
    # Make sure target folder exists
    os.makedirs(target_folder, exist_ok=True)
    
    # Move the file
    if os.path.exists(source_file):
        shutil.copy(source_file, target_folder)
        print(f"✅ Moved {dataset_name} → {domain_folder}")
    else:
        print(f"⚠️ File not found: {dataset_name}")

print("\nAll datasets moved to their Data Mesh domain folders!")
```

```python

for old_name, new_name in dataset_domain_map.items():
    for domain_folder in os.listdir(DATA_MESH_PATH):
        old_path = os.path.join(DATA_MESH_PATH, domain_folder, old_name)
        if os.path.exists(old_path):
            # Ensure new name ends with .csv
            if not new_name.endswith(".csv"):
                new_name += ".csv"
            new_path = os.path.join(DATA_MESH_PATH, domain_folder, new_name)
            os.rename(old_path, new_path)
            print(f"Renamed: {old_name} → {new_name} in {domain_folder}")
```
