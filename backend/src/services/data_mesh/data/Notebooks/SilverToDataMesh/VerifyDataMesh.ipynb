```python
import os
import pandas as pd

data_mesh_path = "../../Data_Mesh_Domains"

# Step 1: List domains and files
for domain in os.listdir(data_mesh_path):
    domain_path = os.path.join(data_mesh_path, domain)
    print(f"\nDomain: {domain}")
    for file in os.listdir(domain_path):
        file_path = os.path.join(domain_path, file)
        try:
            df = pd.read_csv(file_path)
            print(f"{file}: {df.shape} rows x cols")
            
            # --- New: check for null values ---
            null_counts = df.isna().sum()
            total_nulls = null_counts.sum()
            if total_nulls > 0:
                print(f"⚠️ Null values detected ({total_nulls} missing):")
                print(null_counts[null_counts > 0])
            else:
                print(" No null values detected")
            
        except Exception as e:
            print(f"{file}: Failed to read - {e}")
```

```python
# ================================
# Step 2: Load domains into dictionary
# ================================
domains = [
    "customer_domain", "product_domain", "sales_domain",
    "shop_domain", "users_domain", "user_preferences_domain", "engagement_domain"
]

domain_data = {}

print("\n📥 Loading CSVs into memory...\n")
for domain in domains:
    folder_path = os.path.join(data_mesh_path, domain)
    csv_files = [f for f in os.listdir(folder_path) if f.endswith(".csv")]
    
    if csv_files:
        df = pd.read_csv(os.path.join(folder_path, csv_files[0]))
        domain_data[domain] = df
        print(f"{domain}: {df.shape[0]} rows x {df.shape[1]} cols")
        
        # Null check
        null_counts = df.isna().sum()
        if null_counts.sum() > 0:
            print(f"⚠️ {domain} has missing values:")
            print(null_counts[null_counts > 0])
        else:
            print(f"✅ {domain} has no missing values")
    else:
        print(f"No CSV found in {domain}")
```

```python
# ================================
# Step 3: Generate metadata
# ================================
metadata = []

for domain, df in domain_data.items():
    pk_candidate = df.columns[0]  # assuming first column as PK
    metadata.append({
        "domain_name": domain,
        "rows": df.shape[0],
        "columns": df.shape[1],
        "primary_key": pk_candidate,
        "nullable_columns": df.isna().sum().to_dict(),
        "data_types": df.dtypes.apply(lambda x: str(x)).to_dict()
    })
metadata_df = pd.DataFrame(metadata)
```

```python
import json
#================================
# Step 4: Save metadata (CSV & JSON)
# ================================
metadata_path = os.path.join(data_mesh_path, "metadata")
os.makedirs(metadata_path, exist_ok=True)

# CSV
metadata_csv_file = os.path.join(metadata_path, "data_mesh_metadata.csv")
metadata_df.to_csv(metadata_csv_file, index=False)
print(f"\n✅ Metadata CSV saved at: {metadata_csv_file}")

# JSON
metadata_json_file = os.path.join(metadata_path, "data_mesh_metadata.json")
with open(metadata_json_file, "w") as f:
    json.dump(metadata, f, indent=4)
print(f"✅ Metadata JSON saved at: {metadata_json_file}")
```
